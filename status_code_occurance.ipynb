**Question:**  
You have a list of API request logs, where each log is a dictionary:
```python
logs = [
    {"endpoint": "/users", "status": 200},
    {"endpoint": "/orders", "status": 500},
    {"endpoint": "/users", "status": 200},
    {"endpoint": "/login", "status": 401},
    {"endpoint": "/orders", "status": 200},
]
```

Write a Python function that returns how many times each status code appears, e.g. `{200: 3, 500: 1, 401: 1}`.

<h2>Process</h2>

I would create a data structure to store my final answer:
- <u>Empty dictionary</u>: `solution = {}`

I would then iterate over the logs array. For each status value, I would check if that value exists in the solution keys
- If it exists, I would increment the value of that key in the dictionary
- If it doesnt, I would create a key  and assign the value of 1


## Brute Force Solution

````python
def occurance(logs):
    # Empty dictionary containing
    solution = {}
    for log in logs:
        statusCode = log['status']
    
        if statusCode in solution.keys():
            solution[statusCode] += 1
        else:
            solution[statusCode] = 1
        
    return solution
````

````python
logs = [
    {"endpoint": "/users", "status": 200},
    {"endpoint": "/orders", "status": 500},
    {"endpoint": "/users", "status": 200},
    {"endpoint": "/login", "status": 401},
    {"endpoint": "/orders", "status": 200},
]
````

````python
print(occurance(logs))
````

## Tighten Up The Solution

Python has a library that handles this called Counter

````python
from collections import Counter

def updated_occurrance(logs):
    return dict(Counter(log["status"] for log in logs))
````

````python
print(updated_occurrance(logs))
````
